## 3. Data Download and Preparation

Now we'll download the ASL-MNIST dataset from Kaggle. The data comes in CSV files, so we'll need to:

1. Download and unzip the dataset.
2. Create directories to store the train and test images.
3. Read the CSV files using pandas.
4. Convert the pixel data in each row into a 28x28 JPG image and save it to the appropriate directory.

```python
!kaggle datasets download -d datamunge/sign-language-mnist
!unzip sign-language-mnist.zip
```

```python
import os
from pathlib import Path
import pandas as pd
from PIL import Image
import numpy as np
from tqdm.notebook import tqdm

data_path = Path('asl_mnist_data')
train_images_dir = data_path / 'train'
test_images_dir = data_path / 'test'

# Create directories for the images
train_images_dir.mkdir(parents=True, exist_ok=True)
test_images_dir.mkdir(parents=True, exist_ok=True)

train_df = pd.read_csv('sign_mnist_train/sign_mnist_train.csv')
test_df = pd.read_csv('sign_mnist_test/sign_mnist_test.csv')

def process_and_save_images(df, image_dir, split_name):
    print(f'Processing and saving {split_name} images...')
    for index, row in tqdm(df.iterrows(), total=df.shape[0]):
        label = row['label']
        pixels = row[1:].values.reshape(28, 28).astype(np.uint8)
        image = Image.fromarray(pixels)
        image.save(image_dir / f'{split_name}_{index}.jpg')

process_and_save_images(train_df, train_images_dir, 'train')
process_and_save_images(test_df, test_images_dir, 'test')
```

## 4. FiftyOne Dataset Creation

With our images saved as JPG files, we can now create a FiftyOne dataset. This will allow us to easily visualize, query, and analyze our data.

```python
import fiftyone as fo

# Create or load the FiftyOne dataset
if fo.dataset_exists("asl-mnist"):
    asl_dataset = fo.load_dataset("asl-mnist")
    asl_dataset.delete()
asl_dataset = fo.Dataset("asl-mnist")
asl_dataset.persistent = True

# ASL alphabet mapping (excluding J=9 and Z=25 which require motion)
asl_labels = {
    0: 'A', 1: 'B', 2: 'C', 3: 'D', 4: 'E', 5: 'F', 6: 'G', 7: 'H', 8: 'I',
    10: 'K', 11: 'L', 12: 'M', 13: 'N', 14: 'O', 15: 'P', 16: 'Q', 17: 'R',
    18: 'S', 19: 'T', 20: 'U', 21: 'V', 22: 'W', 23: 'X', 24: 'Y'
}

def create_fiftyone_samples(df, image_dir, split_name):
    samples = []
    for index, row in tqdm(df.iterrows(), total=df.shape[0]):
        label_idx = row['label']
        image_path = image_dir / f'{split_name}_{index}.jpg'

        label_letter = asl_labels.get(label_idx, f'unknown_{label_idx}')

        sample = fo.Sample(
            filepath=str(image_path),
            tags=[split_name],
            ground_truth=fo.Classification(label=label_letter)
        )
        samples.append(sample)
    return samples

train_samples = create_fiftyone_samples(train_df, train_images_dir, 'train')
test_samples = create_fiftyone_samples(test_df, test_images_dir, 'test')

asl_dataset.add_samples(train_samples)
asl_dataset.add_samples(test_samples)

print(f"Dataset created with {len(asl_dataset)} samples.")
```

## Step 6: Save the Dataset Locally
To save our FiftyOne dataset for future use or for sharing, we can export it locally. The `dataset.export()` method allows us to save the dataset in various formats, as detailed in the [FiftyOne documentation on exporting datasets](https://docs.voxel51.com/user_guide/export_datasets.html). In this case, we will save it in the `fiftyone` format, which preserves the full FiftyOne dataset structure.



```python
# Define the export directory
export_dir = "/gdrive/MyDrive/fiftyone-saved-work/asl-mnist"
os.makedirs(export_dir, exist_ok=True)

# Specify the format and export the dataset
asl_dataset.export(
    export_dir=export_dir,
    dataset_type=fo.types.FiftyOneDataset,
    label_field="ground_truth"
)

print(f"Dataset exported to: {export_dir}")
```

## Step 7 (optional): Push the Dataset to HuggingFace Hub



This optional step demonstrates how to share your FiftyOne dataset by pushing it to the Hugging Face Hub. The Hugging Face Hub is a platform for sharing and collaborating on machine learning models and datasets.

To push the dataset, we begin by importing the `push_to_hub` function from `fiftyone.utils.huggingface`. Before pushing, you need to authenticate with Hugging Face. This is done by retrieving your Hugging Face API token, which should be securely stored in Colab's secrets manager. This token is then set as an environment variable, allowing the `push_to_hub` function to authenticate and upload the dataset to your account on the Hub.



```python
import os
from google.colab import userdata
os.environ["HF_TOKEN"] = userdata.get('HF_TOKEN')
```

Finally, we call the `push_to_hub` function. We provide the FiftyOne dataset object (`asl_dataset`), the desired name for the dataset on the Hub (`"American-Sign-Language-MNIST"`), and specify the repository type as `"dataset"` and the license as `"mit"`. We also set `exist_ok=True` to allow overwriting if a dataset with the same name already exists. The `chunk_size` parameter is used to control how many samples are uploaded in each batch. Once this cell runs successfully, your ASL-MNIST dataset will be available on the Hugging Face Hub under your account, making it easy to share and collaborate with others.



```python
from fiftyone.utils.huggingface import push_to_hub

push_to_hub(asl_dataset,
            "American-Sign-Language-MNIST",
            repo_type="dataset",
            license="mit",
            exist_ok=True,
            chunk_size=100)
```
